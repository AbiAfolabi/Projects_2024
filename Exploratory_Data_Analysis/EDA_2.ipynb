# Lab Activity One: Guided Data Cleaning

##### It is generally a good idea to make a copy of the master dataset for cleaning so you can always go back if ever needed.

```python
# making a copy of the data frame

df_clean = df1.copy()
```

##### [A] Drop the `'Unnamed: 0'` column

```python
# removing unwanted 'Unnamed' column
df_clean = df_clean.drop(columns=['Unnamed: 0'])

# confirming that 'Unnamed' column is no longer in the dataframe
df_clean.head()
```

```python
# confirming the number of columns has reduced by 1 after removing 'Unnamed' column

df_clean.shape
```

##### [A] Fix the data in the `'Gender'` column.
> Hint: make it so there are 2 unique entries, `'male'` and `'female'`.

```python
# check the unique entries for 'Gender' column

df_clean['Gender'].unique()
```

```python
# rename 'Male' to 'male' and 'fmeale' to 'female'

df_clean['Gender'] = df_clean['Gender'].replace('Male', 'male')
df_clean['Gender'] = df_clean['Gender'].replace('fmeale', 'female')
```

Check the 'Gender' column again

```python
# check the unique entries for 'Gender' column after cleaning

df_clean['Gender'].unique()

# now there are 2 unique entries: 'male' and 'female'
```

##### [A] Fix the data in the `'Mode'` column.
> Hint: Check for lower and upper case entries

```python
# check the unique entries for 'Mode' column

df_clean['Mode'].unique()
```

```python
# change upper case F to lower case

df_clean['Mode'].replace('F', 'f', inplace= True)
```

```python
# check the unique entries for 'Mode' column after cleaning
df_clean['Mode'].unique()
```

### Setting The Correct Data Type

##### [A] Change the `'H'` column to numeric data `type(float64)` instead of `'object'`.

```python
# convert data type of 'H' columnn from 'object' to 'numeric'

df_clean['H'] = pd.to_numeric(df_clean['H'], errors='coerce')
```

*Check* the 'H' column again

```python
df_clean['H'].dtypes
```

##### [A] Convert `'Examination date'` column to `datetime` type.

```python
# convert data type of 'Examination Date' columnn is in 'datetime'

df_clean['Examination Date'] = pd.to_datetime(df_clean['Examination Date'])
```

```python
# confirming that the data types for columns 'H' and 'Examination Date' has changed to 'float64' and 'datetime' respectively

df_clean.dtypes
```

##### Using the `gender_type` key we set the `'Gender'` column as categorical data.

```python
gender_type = pd.CategoricalDtype(categories=["female", "male"])

df_clean["Gender"] = df_clean["Gender"].astype(gender_type)
```

```python
# checking that Gender column data type is now category

df_clean["Gender"].dtypes
```

##### [A] Replicate how we changed the gender column to categorical but this time for the `'Mode'` column.

```python
# Convert 'Mode' column to categorical using the defined mode_type

mode_type = pd.CategoricalDtype(categories=['d', 'e', 'f', 'g', '_', 'h', 'a'])
df_clean["Mode"] = df_clean["Mode"].astype(mode_type)
```

```python
# checking that the data type for 'Mode' column has changed from object to category
df_clean["Mode"].dtypes
```

##### [A] Again, change the `'Diagnosis'` column to categorical replicating above the example above.

```python
# making a list of unique entries for 'Diagnosis' column

df_clean['Diagnosis'].unique()
```

```python
# Convert 'Diagnosis' column to categorical using the defined diagnosis_type

diagnosis_type = pd.CategoricalDtype(categories=['No', 'Yes', '_'])
df_clean["Diagnosis"] = df_clean["Diagnosis"].astype(diagnosis_type)
```

```python
# checking that the data type for 'Diagnosis' column has changed from object to category

df_clean["Diagnosis"].dtypes
```

```python
# checking that the data type for all column to confirm all dtype changes made to Examination Date, Gender,, Mode, H, and Diagnosis columns are effected

df_clean.dtypes
```

### Duplicate Entries
- Duplicate entities in a dataset is not good to have as it  can use overfit and is redundant information

##### [A] Check for duplicate entries and delete them

```python
duplicate_rows = df_clean.duplicated()
```

```python
display(duplicate_rows)
```

before using tilde operator
True = Duplicated, False = not duplicated

After using tilde operator:
False = Duplicated, Trues = Not _ duplicated

```python
# removing duplicates from dataset

df_clean = df_clean[~duplicate_rows]
```

```python
# displaying data rows that are unique after removing duplicates

df_clean
```

```python
# checking the shape of the dataframe after removing duplicates

df_clean.shape
```
