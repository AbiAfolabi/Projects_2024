# **Data** **Exploration**

```
# Clean Titanc Data which columns you think are not needed at all if you think there are unnecessary columns
df_clean = df.drop(['PassengerId'], axis=1)
df_clean.head(5)
```

```
# checking the shape of the clean dataframe to show the number of columns has reduced by 1
df_clean.shape
```

```
# Check how many missing values are there in total in the dataset
df_clean.isnull().sum().sum()
```

```
# Check how many missing values are there in in each column
df_clean.isnull().sum()
```

```
# Checking how many missing values are in the Age column
df_clean['Age'].isnull().sum()
```

```
#Drop all the rows which contain NaN values only in the 'Cabin' column. Be sure to only drop the rows with missing values in the 'Cabin' column and no other column. Remember to use the df_clean dataframe
# making a copy of the clean dataframe
df_clean_cabin = df_clean.copy()

# dropping the rows with NaN values in the 'Cabin' column
df_clean_cabin = df_clean_cabin.dropna(subset=['Cabin'])

# checking the number of missing values in the 'Cabin' column after dropping the rows
print(f"Number of missing values in Cabin after removing Nan:", df_clean_cabin['Cabin'].isnull().sum())

# Check the shape of data after removing NaN rows for the "Cabin" column
print(f"Shape of df_clean after removing NaN from Cabin column:", df_clean_cabin.shape)
```

```
# checking the first five rows of the clean dataframe to show index
df_clean_cabin.head()
```

```
#Use reset index to reset the index after removing the rows

df_clean_cabin = df_clean_cabin.reset_index(drop=True)
df_clean_cabin.head(5)
```

```
# Using df_clean database, create a temporary dataframe and replace all the missing values

# creating a temporary copy of the clean dataframe
df_clean_temp = df_clean.copy()

# in the 'Age' column with the mean of the 'Age' column. Also show the mean value for the Age.
df_clean_temp['Age'].fillna(df_clean_temp['Age'].mean(), inplace=True)

print(f"Mean value of Age:", df_clean_temp['Age'].mean())

# creating a copy of the dataframe after replacing nulls in Age column with the mean value
df_clean_age = df_clean_temp.copy()
```

```
#using df_clean_age dataframe, Use NOCB to fill the NaN values of the entire dataset.

df_clean_age.isnull().sum()
```

```
# filling NaN values in df_clean_age using the NOCB method
df_clean_age = df_clean_age.fillna(method='bfill')
```

```
# checking the NaN values have now been filled with the NOCB method
df_clean_age.isnull().sum()
```
